## macommunaute processing to get pages

### Get all PDF files from the directory

```python
from pathlib import Path

doc_path2 = path2 = "../data/raw/macommunaute/en/"
output_dir2 = "../data/processed/macommunaute"

# Get all PDF files in the directory
pdf_files = list(Path(doc_path2).glob("*.pdf"))
print(f"Found {len(pdf_files)} PDF files:")
for pdf in pdf_files:
    print(f"  - {pdf.name}")
```

### Process each PDF file and extract pages

```python
# Process all PDF files - same format as first part
os.makedirs(output_dir2, exist_ok=True)

for pdf_file in pdf_files:
    print(f"Processing: {pdf_file.name}")
    doc = fitz.open(str(pdf_file))
    print(f"  Pages: {len(doc)}")
    
    # Extract pages in same format as first part
    pages = []
    for i in range(len(doc)):
        text = doc[i].get_text("text")
        pages.append({"page_number": i+1, "text": text})
    
    # Save each PDF to its own JSON file
    json_filename = pdf_file.stem + ".json"
    output_path = os.path.join(output_dir2, json_filename)
    
    with open(output_path, "w", encoding="utf-8") as f:
        json.dump(pages, f, ensure_ascii=False, indent=2)
    
    print(f"  ✓ Saved to: {json_filename}\n")
    doc.close()

print(f"✓ Processing complete! All PDFs saved to: {output_dir2}")
```
